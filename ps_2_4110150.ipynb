```python
import runpy
n = 9
A = np.ones((n, n)) + 2 * np.eye(n)
b = np.zeros(n)
for i in range(n):
    for j in range(n):
        b[i] = b[i] + (j + 1) * A[i, j]


gaussian_elimination = runpy.run_path(
    "ps_2_ex 4b_4110150.py"
)["gaussian_elimination"]


x = gaussian_elimination(A, b)

print("Matrix A:")
print(A)

print("Vector b:")
print(b)

print("Solution vector x:")
print(x)

print("Verification:", np.allclose(A @ x, b))
```
